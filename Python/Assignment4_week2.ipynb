```python
class OddEven:
    def OddEven():
        num=int(input("Enter any number: "))
        if(num%2==1):
            print(num," is odd number")
        else:
            print(num," is even number")

OddEven.OddEven()
            
```

```python
class EligiblityForMarriage:
    def Eligible():
        gender=input("Your Gender : ")
        age=int(input("Your Age : "))
        if (gender=='Male'):
                if (age>20):
                    print("Eligible for Marriage")
                else:
                    print("Not Eligible for Marriage")
        elif (gender=='Female'):
            if (age>17):
                print("Eligible for Marriage")
            else:
                print("Not Eligible for Marriage")
        else:
            print("Enter the correct gender Male or Female")

EligiblityForMarriage.Eligible()
```

```python
class FindPercent:
    def percentage():
        total=0
        for i in range(1,6):
            i=int(input(f"Subject {i}: "))
            total=total+i
        print("Total : ",total)
        percentage=total/5
        print(f"Percentage : {percentage}")
        
FindPercent.percentage()
```
